```python
import math

def q_solve(a, b, c):

    if a == 0:
        print("Not a quadratic equation")
        return

    D = b**2 - 4*a*c

    if D > 0:
        x1 = (-b + math.sqrt(D)) / (2*a)
        x2 = (-b - math.sqrt(D)) / (2*a)

        print("Two real solutions:")
        print("x1 =", x1)
        print("x2 =", x2)

    elif D == 0:
        x = -b / (2*a)
        print("One real solution:")
        print("x =", x)

    else:
        print("no real solution")
```

```python
q_solve(1, -5, 6)
```

```python
q_solve(1, -4, 4)
```

```python
q_solve(1, 2, 5)
```
